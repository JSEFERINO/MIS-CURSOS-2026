### $\blacksquare$ **Ejemplo: Simulación de Lanzamiento de dos dados distinguibles**

```python
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML
import numpy as np

# Función para lanzar dos dados
def lanzar_dos_dados():
    return np.random.randint(1, 7, size=2)

# Inicialización de variables para contar los resultados
resultados = {(i, j): 0 for i in range(1, 7) for j in range(1, 7)}
historial_resultados = []

# Función para actualizar los resultados, la tabla y el gráfico en cada fotograma de la animación
def actualizar(i):
    resultado = lanzar_dos_dados()
    resultados[tuple(resultado)] += 1
    historial_resultados.append(resultado)

    ejes[0].clear()
    ejes[0].text(0.5, 0.5, str(resultado), fontsize=20, ha='center', va='center')

    ejes[1].clear()
    x_labels = [f"{i}-{j}" for i in range(1, 7) for j in range(1, 7)]
    ejes[1].bar(x_labels, [resultados[(i, j)] for i in range(1, 7) for j in range(1, 7)], color='blue')
    ejes[1].set_xticklabels(x_labels, rotation=45, ha='right')
    ejes[1].set_ylim(0, max(resultados.values()) + 1)


# Configuración inicial del gráfico
fig, ejes = plt.subplots(1, 2, figsize=(10, 5))

# Configuración de la animación
ani = animation.FuncAnimation(fig, actualizar, frames=29, interval=500, repeat=False)

# Muestra la animación en el notebook
HTML(ani.to_html5_video())

```

## $\blacksquare$ **Ejemplo. Simulando el Lanzamiento de dos dados y obtener el total**

```python
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML
import numpy as np

# Función para lanzar dos dados y obtener el total
def lanzar_dados_y_obtener_total():
    return np.random.randint(1, 7, size=2).sum()

# Inicialización de variables para contar los resultados
resultados = {i: 0 for i in range(2, 13)}
historial_resultados = []

# Función para actualizar los resultados, la tabla y el gráfico en cada fotograma de la animación
def actualizar(i):
    total = lanzar_dados_y_obtener_total()
    resultados[total] += 1
    historial_resultados.append(total)

    ejes[0].clear()
    ejes[0].text(0.5, 0.5, str(total), fontsize=30, ha='center', va='center')

    ejes[1].clear()
    x_labels = list(range(2, 13))
    ejes[1].bar(x_labels, [resultados[i] for i in x_labels], color='blue')
    ejes[1].set_xticks(x_labels)
    ejes[1].set_ylim(0, max(resultados.values()) + 1)


# Configuración inicial del gráfico
fig, ejes = plt.subplots(1, 2, figsize=(10, 5))

# Configuración de la animación
ani = animation.FuncAnimation(fig, actualizar, frames=49, interval=500, repeat=False)

# Muestra la animación en el notebook
HTML(ani.to_html5_video())
```

```python
from matplotlib import pyplot as plt
from matplotlib import animation
import numpy as np

# Función para lanzar monedas
def lanzar_monedas():
    return np.random.choice(['C', 'S'], size=2)

# Variables
resultados = {'C-C': 0, 'C-S': 0, 'S-C': 0, 'S-S': 0}
historial_resultados = []

# Colores para cada resultado
colores = {
    'C-C': 'blue',
    'C-S': 'red',
    'S-C': 'green',
    'S-S': 'orange'
}

# Configuración inicial
fig, ejes = plt.subplots(1, 2, figsize=(10, 5))

# Función de actualización
def actualizar(i):
    resultados_lanzamiento = '-'.join(lanzar_monedas())
    resultados[resultados_lanzamiento] += 1
    historial_resultados.append(resultados_lanzamiento)

    # Mostrar resultado actual
    ejes[0].clear()
    ejes[0].text(
        0.5,
        0.5,
        resultados_lanzamiento,
        fontsize=30,
        ha='center',
        va='center',
        color=colores[resultados_lanzamiento]  # Color dinámico
    )

    ejes[0].set_axis_off()
    ejes[0].set_title(f"Lanzamiento {i+1}")

    # Gráfico de barras
    ejes[1].clear()

    ejes[1].bar(
        resultados.keys(),
        resultados.values(),
        color=[colores[k] for k in resultados.keys()]
    )

    ejes[1].set_ylim(0, 40)
    ejes[1].set_title("Frecuencia de resultados")

# Crear animación
ani = animation.FuncAnimation(
    fig,
    actualizar,
    frames=100,
    interval=1000,
    repeat=False
)

# Guardar como GIF
gif_path = "simulacion_monedas.gif"

ani.save(gif_path, writer='pillow', fps=1)

plt.close(fig)

print(f"GIF guardado en: {gif_path}")
```

```python
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML
import numpy as np

# Función para lanzar una moneda
def lanzar_moneda():
    return np.random.choice(['CARA', 'SELLO'])

# Función para lanzar un dado
def lanzar_dado():
    return np.random.randint(1, 7)

# Inicialización de variables para contar los resultados
resultados_moneda = {'CARA': 0, 'SELLO': 0}
resultados_dado = {i: 0 for i in range(1, 7)}
historial_resultados = []

# Función para actualizar los resultados, la tabla y el gráfico en cada fotograma de la animación
def actualizar(i):
    resultado_moneda = lanzar_moneda()
    resultado_dado = lanzar_dado()
    resultados_moneda[resultado_moneda] += 1
    resultados_dado[resultado_dado] += 1
    historial_resultados.append((resultado_moneda, resultado_dado))

    ejes[0].clear()
    ejes[0].text(0.5, 0.5, f"{resultado_moneda} - {resultado_dado}", fontsize=15, ha='center', va='center')

    ejes[1].clear()
    ejes[1].bar(resultados_moneda.keys(), resultados_moneda.values(), color=['blue', 'red'])

    ejes[2].clear()
    ejes[2].bar(resultados_dado.keys(), resultados_dado.values(), color='green')
    ejes[2].set_xticks(list(resultados_dado.keys()))

# Configuración inicial del gráfico
fig, ejes = plt.subplots(1, 3, figsize=(10, 5))

# Configuración de la animación
ani = animation.FuncAnimation(fig, actualizar, frames=19, interval=500, repeat=False)

# Muestra la animación en el notebook
HTML(ani.to_html5_video())
```

```python
from matplotlib import pyplot as plt
from matplotlib import animation
import numpy as np

# Parámetros del problema
s = 10.0          # Lado del cuadrado
r = 3.0           # Radio del círculo
centro_x = s / 2  # Centro del círculo (x)
centro_y = s / 2  # Centro del círculo (y)

# Parámetros de simulación
num_frames = 150
interval_ms = 80

# Estadísticas
aciertos = 0
total = 0

# Figura
fig, ax = plt.subplots(figsize=(7, 7))

def actualizar(frame):
    global aciertos, total

    ax.clear()

    # Generar punto aleatorio en el cuadrado [0, s] x [0, s]
    x = np.random.uniform(0, s)
    y = np.random.uniform(0, s)
    total += 1

    # Verificar si está dentro del círculo
    distancia = np.sqrt((x - centro_x)**2 + (y - centro_y)**2)
    esta_dentro = distancia <= r

    if esta_dentro:
        aciertos += 1

    prob_estimada = aciertos / total
    prob_teorica = np.pi * r**2 / s**2

    # Dibujar cuadrado
    cuadrado = plt.Rectangle((0, 0), s, s, fill=False, edgecolor='black', linewidth=2)
    ax.add_patch(cuadrado)

    # Dibujar círculo
    circulo = plt.Circle((centro_x, centro_y), r, color='blue', alpha=0.3, label=f'Círculo (r={r} cm)')
    ax.add_patch(circulo)

    # Dibujar punto
    color = 'green' if esta_dentro else 'red'
    ax.scatter(x, y, s=100, color=color, zorder=5, edgecolors='black', linewidth=1)

    # Textos informativos
    ax.text(0.5, s + 0.8, f"Punto: ({x:.2f}, {y:.2f}) cm", fontsize=10, ha='center')
    ax.text(0.5, s + 0.3, f"¿Dentro del círculo?: {'SÍ' if esta_dentro else 'NO'}",
            fontsize=11, color=color, ha='center', weight='bold')
    ax.text(0.5, s - 0.3, f"Probabilidad estimada: {prob_estimada:.4f}",
            fontsize=10, ha='center', style='italic')
    ax.text(0.5, s - 0.8, f"Probabilidad teórica: {prob_teorica:.4f} = π·{r}²/{s}²",
            fontsize=10, ha='center', style='italic')

    # Configuración
    ax.set_xlim(-1, s + 1)
    ax.set_ylim(-1, s + 1.5)
    ax.set_xlabel("x (cm)", fontsize=11)
    ax.set_ylabel("y (cm)", fontsize=11)
    ax.set_title(f"Simulación: Punto aleatorio en cuadrado de lado {s} cm\ncon círculo inscrito de radio {r} cm",
                 fontsize=12, weight='bold')
    ax.set_aspect('equal')
    ax.grid(True, linestyle='--', alpha=0.3)
    ax.legend(loc='upper right')

# Crear animación
ani = animation.FuncAnimation(
    fig,
    actualizar,
    frames=num_frames,
    interval=interval_ms,
    repeat=False
)

# Guardar GIF
gif_path = "simulacion_punto_en_cuadrado_con_circulo.gif"
ani.save(gif_path, writer='pillow', fps=1000/interval_ms)

plt.close(fig)

print("✅ GIF creado correctamente")
print(f"📁 Ruta: {gif_path}")
print(f"🎲 Probabilidad teórica: π·{r}²/{s}² = {np.pi * r**2 / s**2:.6f}")
print(f"📊 Fórmula: P(A) = πr²/s² = {np.pi:.4f}·{r**2}/{s**2} = {np.pi * r**2}/{s**2}")
```

```python
from matplotlib import pyplot as plt
from matplotlib import animation
import numpy as np

# Parámetros del problema
p_verdadera = 0.5      # Probabilidad de éxito (cara)
epsilon = 0.05         # Margen de error
n_total = 500          # Número total de lanzamientos
intervalo_confianza_inferior = p_verdadera - epsilon
intervalo_confianza_superior = p_verdadera + epsilon

# Parámetros de animación
num_frames = n_total   # Un frame por lanzamiento
interval_ms = 50       # Milisegundos entre frames

# Estadísticas
exitos = 0
frecuencias = []
ensayos = []

# Figura con dos subplots
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 8))
fig.suptitle('Ley de los Grandes Números (Jacob Bernoulli, 1713)',
             fontsize=14, weight='bold')

def actualizar(frame):
    global exitos, frecuencias, ensayos

    # Limpiar ejes
    ax1.clear()
    ax2.clear()

    # Simular un lanzamiento (éxito = cara)
    es_exito = np.random.random() < p_verdadera
    if es_exito:
        exitos += 1

    n = frame + 1  # Número de ensayos hasta ahora
    frecuencia = exitos / n

    # Guardar historial
    frecuencias.append(frecuencia)
    ensayos.append(n)

    # === PRIMER GRÁFICO: Evolución de la frecuencia ===
    ax1.plot(ensayos, frecuencias, 'b-', linewidth=2, label='Frecuencia relativa (k/n)')
    ax1.axhline(y=p_verdadera, color='green', linestyle='-',
                linewidth=2, label=f'p = {p_verdadera} (verdadera)')
    ax1.axhline(y=intervalo_confianza_inferior, color='orange',
                linestyle='--', linewidth=1.5, alpha=0.7)
    ax1.axhline(y=intervalo_confianza_superior, color='orange',
                linestyle='--', linewidth=1.5, alpha=0.7)

    # Rellenar banda de confianza
    ax1.fill_between(ensayos,
                     intervalo_confianza_inferior,
                     intervalo_confianza_superior,
                     alpha=0.2, color='orange',
                     label=f'Banda p ± ε (ε={epsilon})')

    # Punto actual
    ax1.scatter(n, frecuencia, color='red', s=80, zorder=5)

    # Configuración primer gráfico
    ax1.set_xlim(0, n_total)
    ax1.set_ylim(0, 1)
    ax1.set_xlabel('Número de lanzamientos (n)', fontsize=10)
    ax1.set_ylabel('Frecuencia relativa (k/n)', fontsize=10)
    ax1.set_title('Convergencia de la frecuencia relativa', fontsize=11, weight='bold')
    ax1.grid(True, alpha=0.3)
    ax1.legend(loc='upper right', fontsize=8)

    # === SEGUNDO GRÁFICO: Barras de error ===
    error_absoluto = abs(frecuencia - p_verdadera)
    dentro_banda = error_absoluto < epsilon

    # Crear gráfico de barras para el error
    colores_barra = ['green' if dentro_banda else 'red']
    ax2.bar(0, error_absoluto, color=colores_barra[0], alpha=0.7, width=0.6)
    ax2.axhline(y=epsilon, color='orange', linestyle='--',
                linewidth=2, label=f'ε = {epsilon}')

    # Configuración segundo gráfico
    ax2.set_xlim(-0.5, 0.5)
    ax2.set_ylim(0, 1)
    ax2.set_xticks([])
    ax2.set_ylabel('Error absoluto |k/n - p|', fontsize=10)
    ax2.set_title(f'Error: {"DENTRO" if dentro_banda else "FUERA"} de la banda (n={n})',
                  fontsize=11, weight='bold', color=colores_barra[0])
    ax2.grid(True, alpha=0.3, axis='y')
    ax2.legend(loc='upper right', fontsize=9)

    # Añadir texto informativo
    info_texto = f"""
    Lanzamientos: {n}
    Éxitos (caras): {exitos}
    Frecuencia: {frecuencia:.4f}
    Error: {error_absoluto:.4f}
    Teórico: p = {p_verdadera}
    """
    ax2.text(0.05, 0.95, info_texto, transform=ax2.transAxes,
             fontsize=9, verticalalignment='top',
             bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5),
             family='monospace')

# Crear animación
ani = animation.FuncAnimation(
    fig,
    actualizar,
    frames=num_frames,
    interval=interval_ms,
    repeat=False
)

# Guardar GIF
gif_path = "ley_grandes_numeros_moneda.gif"
ani.save(gif_path, writer='pillow', fps=1000/interval_ms)

plt.close(fig)

print("✅ GIF creado correctamente")
print(f"📁 Ruta: {gif_path}")
print(f"\n🎲 Parámetros de la simulación:")
print(f"   • Probabilidad teórica (p): {p_verdadera}")
print(f"   • Margen de error (ε): {epsilon}")
print(f"   • Total de lanzamientos: {n_total}")
print(f"\n📊 Lo que demuestra el GIF:")
print(f"   • Al principio: grandes fluctuaciones")
print(f"   • A medida que n crece: la frecuencia se acerca a p")
print(f"   • El error |k/n - p| tiende a ser menor que ε")
print(f"\n🎯 Interpretación del Teorema:")
print(f"   lim P(|k/n - p| < ε) = 1  cuando n → ∞")
print(f"   • Para n={n_total}, ya se observa convergencia")
```

```python
from matplotlib import pyplot as plt
from matplotlib import animation
import numpy as np
from IPython.display import Image, display # Importar Image y display

# ==========================================
# PARÁMETROS DEL PROBLEMA
# ==========================================

p_defecto = 0.08       # Probabilidad real de defecto
epsilon = 0.02         # Margen de error permitido
n_total = 500          # Artículos inspeccionados

lim_inf = p_defecto - epsilon
lim_sup = p_defecto + epsilon

# Parámetros de animación
num_frames = n_total
interval_ms = 50

# Estadísticas
defectuosos = 0
proporciones = []
inspecciones = []

# ==========================================
# FIGURA
# ==========================================

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 8))

fig.suptitle(
    'Ley de los Grandes Números aplicada al Control de Calidad',
    fontsize=14,
    weight='bold'
)

# ==========================================
# FUNCIÓN DE ACTUALIZACIÓN
# ==========================================

def actualizar(frame):

    global defectuosos, proporciones, inspecciones

    ax1.clear()
    ax2.clear()

    # Simulación de inspección
    es_defectuoso = np.random.random() < p_defecto

    if es_defectuoso:
        defectuosos += 1

    n = frame + 1

    proporcion = defectuosos / n

    proporciones.append(proporcion)
    inspecciones.append(n)

    # ======================================
    # GRÁFICO 1
    # ======================================

    ax1.plot(
        inspecciones,
        proporciones,
        'b-',
        linewidth=2,
        label='Proporción observada de defectos'
    )

    ax1.axhline(
        y=p_defecto,
        color='green',
        linewidth=2,
        label=f'Probabilidad real = {p_defecto}'
    )

    ax1.axhline(
        y=lim_inf,
        color='orange',
        linestyle='--'
    )

    ax1.axhline(
        y=lim_sup,
        color='orange',
        linestyle='--'
    )

    ax1.fill_between(
        inspecciones,
        lim_inf,
        lim_sup,
        color='orange',
        alpha=0.2,
        label=f'Banda ± ε (ε={epsilon})'
    )

    ax1.scatter(
        n,
        proporcion,
        color='red',
        s=80,
        zorder=5
    )

    ax1.set_xlim(0, n_total)
    ax1.set_ylim(0, 0.30)

    ax1.set_xlabel('Número de artículos inspeccionados')
    ax1.set_ylabel('Proporción de defectuosos')

    ax1.set_title(
        'Estimación de la probabilidad de defecto',
        fontsize=11,
        weight='bold'
    )

    ax1.grid(True, alpha=0.3)
    ax1.legend()

    # ======================================
    # GRÁFICO 2
    # ======================================

    error = abs(proporcion - p_defecto)

    dentro_banda = error < epsilon

    color_barra = 'green' if dentro_banda else 'red'

    ax2.bar(
        0,
        error,
        color=color_barra,
        alpha=0.7,
        width=0.6
    )

    ax2.axhline(
        y=epsilon,
        color='orange',
        linestyle='--',
        linewidth=2,
        label=f'ε = {epsilon}'
    )

    ax2.set_xlim(-0.5, 0.5)
    ax2.set_ylim(0, 0.15)

    ax2.set_xticks([])

    ax2.set_ylabel('|p̂ - p|')

    ax2.set_title(
        f'Error de estimación: {"DENTRO" if dentro_banda else "FUERA"} de la banda',
        fontsize=11,
        weight='bold',
        color=color_barra
    )

    ax2.grid(True, alpha=0.3)

    info = f"""
Artículos inspeccionados: {n}

Defectuosos encontrados: {defectuosos}

Proporción observada:
p̂ = {proporcion:.4f}

Probabilidad real:
p = {p_defecto}

Error:
|p̂ - p| = {error:.4f}
"""

    ax2.text(
        0.05,
        0.95,
        info,
        transform=ax2.transAxes,
        fontsize=9,
        verticalalignment='top',
        bbox=dict(boxstyle='round',
                  facecolor='lightyellow',
                  alpha=0.7),
        family='monospace'
    )

    ax2.legend()

# ==========================================
# CREAR ANIMACIÓN
# ==========================================

ani = animation.FuncAnimation(
    fig,
    actualizar,
    frames=num_frames,
    interval=interval_ms,
    repeat=False
)

gif_path = "estimacion_probabilidad_defectos.gif"

ani.save(
    gif_path,
    writer="pillow",
    fps=1000/interval_ms
)

plt.close(fig)

print("✅ GIF creado correctamente")
print(f"📁 Archivo: {gif_path}")
display(Image(filename=gif_path)) # Mostrar el GIF al final
```
